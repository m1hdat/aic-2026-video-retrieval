# CLIP Retrieval Demo

This notebook demonstrates the retrieval idea before using Milvus.

```python
from pathlib import Path
import sys

PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
sys.path.insert(0, str(PROJECT_ROOT))

from src.config import load_config
from src.dataset_loader import get_image_paths
from src.clip_encoder import ClipEncoder

config = load_config(PROJECT_ROOT / 'configs' / 'config.yaml')
image_paths = get_image_paths(config['data']['image_dir'])[:8]
encoder = ClipEncoder(config['model']['name'], device=config['model'].get('device', 'auto'))
```

```python
import numpy as np

query = 'a child playing outside'
image_embeddings = np.vstack([encoder.encode_image(path) for path in image_paths])
text_embedding = encoder.encode_text(query)
scores = image_embeddings @ text_embedding
top_indices = scores.argsort()[::-1][:5]

[(str(image_paths[index]), float(scores[index])) for index in top_indices]
```
